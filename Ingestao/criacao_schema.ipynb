```python
import pandas as pd
import psycopg2
```

```python
dbname = 'dw_eletronicos'
user = 'postgres'
password = '12345'
host = 'localhost'
port = '5432'

conexao = psycopg2.connect(dbname=dbname, user=user, password=password, host=host, port=port)
cursor = conexao.cursor()


cursor.execute(""" 
                CREATE SCHEMA IF NOT EXISTS raw
               """)
conexao.commit()
cursor.close()
conexao.close()
```
