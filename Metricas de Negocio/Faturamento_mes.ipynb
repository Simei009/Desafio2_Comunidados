```python
import pandas as pd

from sqlalchemy import create_engine

engine = create_engine('postgresql+psycopg2://postgres:12345@localhost:5432/dw_eletronicos')

faturamento_mes = """

                    select dt.nome_mes as mes, dt.ano,
                        round(SUM(valor_total), 2) as faturamento
                        from dw.dim_tempo dt
                        join dw.fato_vendas dv on dt.data = dv.data_pedido
                        group by dt.nome_mes, dt.ano
                        order by dt.ano,
                            case dt.nome_mes
                                when 'Janeiro' then 1
                                when 'Fevereiro' then 2
                                when 'Março' then 3
                                when 'Abril' then 4
                                when 'Maio' then 5
                                when 'Junho' then 6
                                when 'Julho' then 7
                                when 'Agosto' then 8
                                when 'Setembro' then 9
                                when 'Outubro' then 10
                                when 'Novembro' then 11
                                when 'Dezembro' then 12
                            end;


               """

df = pd.read_sql(faturamento_mes, engine)

df.head(30)


```

```python


from sqlalchemy import create_engine

engine = create_engine('postgresql+psycopg2://postgres:12345@localhost:5432/dw_eletronicos')

faturamento_mes = """

                    select dt.nome_mes as mes, dt.ano,
                        round(SUM(valor_total), 2) as faturamento
                        from dw.dim_tempo dt
                        join dw.fato_vendas dv on dt.data = dv.data_pedido
                        group by dt.nome_mes, dt.ano
                        order by dt.ano,
                            case dt.nome_mes
                                when 'Janeiro' then 1
                                when 'Fevereiro' then 2
                                when 'Março' then 3
                                when 'Abril' then 4
                                when 'Maio' then 5
                                when 'Junho' then 6
                                when 'Julho' then 7
                                when 'Agosto' then 8
                                when 'Setembro' then 9
                                when 'Outubro' then 10
                                when 'Novembro' then 11
                                when 'Dezembro' then 12
                            end;


               """

df = pd.read_sql(faturamento_mes, engine)

df.head(30)

```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

plt.bar(df['mes'], df['faturamento'])

plt.title('Faturamento por Mês')
plt.xlabel('Mês')
plt.ylabel('Faturamento')

plt.xticks(rotation=45)

plt.show()
```
