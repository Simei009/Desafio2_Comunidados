```python
import pandas as pd

from sqlalchemy import create_engine

engine = create_engine('postgresql+psycopg2://postgres:12345@localhost:5432/dw_eletronicos')

faturamento_produto = """

                    select dp.produto_id, dp.produto,
                            'R$ ' || sum(fv.valor_total) as faturamento_produto
                        from dw.dim_produto dp
                        join dw.fato_vendas fv on dp.produto_id = fv.produto_id
                        group by dp.produto_id, dp.produto
                        order by faturamento_produto;

               """

df = pd.read_sql(faturamento_produto, engine)

df.head(50)

```
