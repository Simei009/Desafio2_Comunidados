```python
import pandas as pd
from sqlalchemy import create_engine


engine = create_engine('postgresql+psycopg2://postgres:12345@localhost:5432/dw_eletronicos')

faturamento_liquido = """
                                with devolucoes as (
                                    select sum(
                                            case 
                                                when lower(status_devolucao) = 'em analise' then 0
                                                else valor_devolvido
                                                end 
                                    ) as valor_devolucoes
                                    from dw.fato_devolucoes
                                ),
                                faturamento_vendas as (
                                    select sum(
                                            case 
                                                when lower(status_pedido) = 'cancelado' then 0
                                                else(valor_total - frete)
                                                end 
                                    ) as valor_total_vendas
                                    from dw.fato_vendas
                                )
                                select round(valor_total_vendas - valor_devolucoes, 2) as faturamento_liquido
                                    from devolucoes,faturamento_vendas; 
                                """
df = pd.read_sql(faturamento_liquido, engine)
df.head()

```
