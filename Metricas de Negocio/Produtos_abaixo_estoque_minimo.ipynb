```python
import pandas as pd

from sqlalchemy import create_engine

engine = create_engine('postgresql+psycopg2://postgres:12345@localhost:5432/dw_eletronicos')

faturamento_mes = """

                    select produto, categoria, fornecedor, estoque_atual, estoque_minimo
                    from dw.dim_produto
                    where estoque_atual < estoque_minimo
                    order by estoque_atual;


               """

df = pd.read_sql(faturamento_mes, engine)

df.head(30)
```
