```python
import pandas as pd

from sqlalchemy import create_engine

engine = create_engine('postgresql+psycopg2://postgres:XXXXXX@localhost:5432/dw_eletronicos')

faturamento_estado = """
                                select estado,
                                        sum(fv.valor_total) as faturamento_estado
                                    from dw.dim_localidade dl
                                    join dw.fato_vendas fv on dl.id_localidade = fv.id_localidade
                                    group by estado
                                    order by faturamento_estado;

                                """
df = pd.read_sql(faturamento_estado, engine)
##df.head(50)
print(df.to_string(index=False))


```
