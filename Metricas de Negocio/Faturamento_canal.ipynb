```python
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine('postgresql+psycopg2://postgres:12345@localhost:5432/dw_eletronicos')

faturamento_canal_venda = """
                                select dcv.canal_venda,
                                    count(fv.pedido_id) as quantidade_pedido,
                                    sum(fv.valor_total) as faturamento_canal
                                    from dw.dim_canal_venda as dcv
                                    join dw.fato_vendas fv on fv.id_canal_venda = dcv.id_canal_venda
                                    group by canal_venda
                                    order by faturamento_canal;

                                """
df = pd.read_sql(faturamento_canal_venda, engine)
df.head(50)


```
