```python
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine('postgresql+psycopg2://postgres:12345@localhost:5432/dw_eletronicos')

faturamento_forma_pagamento = """
                                select dp.forma_pagamento,
                                count(fv.pedido_id) as quantidade_pedido,
                                sum(fv.valor_total) as faturamento_vendas
                                from dw.dim_pagamento dp
                                join dw.fato_vendas fv on fv.id_pagamento = dp.id_pagamento
                                group by forma_pagamento
                                order by faturamento_vendas;

                                """
df = pd.read_sql(faturamento_forma_pagamento, engine)
df.head(50)

```
