```python
import pandas as pd

from sqlalchemy import create_engine



engine = create_engine('postgresql+psycopg2://postgres:12345@localhost:5432/dw_eletronicos')

faturamento_categoria = """
                                select categoria,
                                    sum(valor_total) as faturamento_categoria
                                from dw.dim_produto dp
                                join dw.fato_vendas fv on dp.produto_id = fv.produto_id
                                group by categoria
                                order by faturamento_categoria;

                                """
df = pd.read_sql(faturamento_categoria, engine)
df.head(50)
```
